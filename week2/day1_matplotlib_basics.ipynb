# Task 1.2: Line Charts

```python
import matplotlib.pyplot as plt

# Base Data
months = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
sales = [120, 135, 148, 162, 155, 170, 185, 178, 195, 210, 225, 240]
expenses = [100, 110, 115, 125, 130, 135, 140, 145, 150, 155, 165, 170]

plt.figure(figsize=(10, 6))

# Plotting with explicit styles from instructions
plt.plot(months, sales, marker='o', color='#0D9488', linewidth=2, linestyle='-', label='Sales')
plt.plot(months, expenses, marker='s', color='#EF4444', linewidth=2, linestyle='--', label='Expenses')

# Customization
plt.title('Monthly Sales vs Expenses', fontsize=16, fontweight='bold')
plt.xlabel('Month', fontsize=12)
plt.ylabel('Amount ($K)', fontsize=12)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)

# Rotate labels to prevent overlap
plt.xticks(rotation=45)
plt.tight_layout()

# show
plt.show()
```

# Task 1.3: Bar Charts

```python
import matplotlib.pyplot as plt

# 1. Base Data
departments = ['IT', 'Sales', 'HR', 'Marketing', 'Finance']
scores = [88, 92, 76, 85, 90]
colors = ['#0D9488', '#1B2A4A', '#F59E0B', '#EF4444', '#10B981']

# Part 1: Vertical Bar Chart
plt.figure(figsize=(8, 5))

# Vertical bars
bars = plt.bar(departments, scores, color=colors, edgecolor='white', linewidth=1.5)

# Add labels on top
for bar, score in zip(bars, scores):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 1,
             str(score), ha='center', fontweight='bold', fontsize=11)

plt.title('Department Performance Scores (Vertical)', fontsize=14, fontweight='bold')
plt.ylabel('Score')
plt.ylim(0, 105)
plt.tight_layout()
plt.show()

# Part 2: Horizontal Bar Chart
plt.figure(figsize=(8, 5))

# Horizontal bars
bars_h = plt.barh(departments, scores, color=colors, edgecolor='white', linewidth=1.5)

# Add labels to the right
for bar, score in zip(bars_h, scores):
    plt.text(bar.get_width() + 1, bar.get_y() + bar.get_height()/2,
             str(score), va='center', ha='left', fontweight='bold', fontsize=11)

plt.title('Department Performance Scores (Horizontal)', fontsize=14, fontweight='bold')
plt.xlabel('Score')
plt.xlim(0, 105)
plt.tight_layout()
plt.show()
```

# Task 1.4: Scatter Plots

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Generate sample data
np.random.seed(42)
hours_studied = np.random.uniform(1, 10, 50)
exam_scores = 40 + 5 * hours_studied + np.random.normal(0, 5, 50)

# 2. Figure setup
plt.figure(figsize=(8, 6))

# 3. Scatter plot with requested modification (Changed cmap to 'plasma')
plt.scatter(hours_studied, exam_scores, c=exam_scores, cmap='plasma',
            s=80, alpha=0.7, edgecolors='white')

# 4. Add trend line
z = np.polyfit(hours_studied, exam_scores, 1)
p = np.poly1d(z)
x_line = np.linspace(1, 10, 100)
plt.plot(x_line, p(x_line), '--', color='red', alpha=0.8, label=f'Trend (slope={z[0]:.1f})')

# 5. Customization
plt.colorbar(label='Score')
plt.title('Study Hours vs Exam Score', fontsize=14, fontweight='bold')
plt.xlabel('Hours Studied')
plt.ylabel('Exam Score')
plt.legend()
plt.tight_layout()

# 6. Show plot (plt.savefig removed)
plt.show()
```

# Task 1.5: Histograms

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Generate sample data (From original file)
np.random.seed(42)
salaries = np.concatenate([
    np.random.normal(50000, 10000, 200),   # Junior
    np.random.normal(80000, 15000, 150),    # Mid
    np.random.normal(120000, 20000, 50),    # Senior
])

mean_salary = np.mean(salaries)
bins_to_try = [10, 20, 50]

# 2. Plotting histograms for each bin size to see the effect
for b in bins_to_try:
    plt.figure(figsize=(10, 5))

    # [✓] Created a histogram showing a salary distribution with dynamic bins
    plt.hist(salaries, bins=b, color='#0D9488', edgecolor='white', alpha=0.7, label='All employees')

    # [✓] Added a mean line with plt.axvline
    plt.axvline(mean_salary, color='#EF4444', linestyle='--', linewidth=2,
                label=f'Mean: ${mean_salary:,.0f}')

    # Customization
    plt.title(f'Salary Distribution (bins={b})', fontsize=14, fontweight='bold')
    plt.xlabel('Annual Salary ($)')
    plt.ylabel('Number of Employees')
    plt.legend(fontsize=11)
    plt.tight_layout()

    # Show plot directly (plt.savefig removed)
    plt.show()
```
