```python
import plotly.express as px
import plotly.graph_objects as go
import pandas as pd
import numpy as np
```

# Task 4.1:  Install Plotly & First Interactive Chart

```python
# Load dataset and filter for 2007
df = px.data.gapminder().query('year == 2007')

# Create interactive scatter plot
fig = px.scatter(
    df, 
    x='gdpPercap', 
    y='lifeExp',
    size='pop', 
    color='continent',
    hover_name='country',
    log_x=True, 
    size_max=60,
    title='World Development in 2007'
)

# Show plot and save as HTML in the current directory
fig.show()
fig.write_html('interactive_scatter.html')
```

# Task 4.2:  Interactive Line & Bar Charts

```python
# Load dataset
gapminder = px.data.gapminder()
countries = ['Saudi Arabia', 'United Arab Emirates', 'Japan', 'Germany', 'Brazil']
df_filtered = gapminder[gapminder['country'].isin(countries)]

# 1. Interactive line chart with range slider
fig = px.line(df_filtered, x='year', y='gdpPercap', color='country',
              title='GDP Per Capita Over Time', markers=True)

# Add range slider and hover effects
fig.update_layout(
    xaxis=dict(rangeslider=dict(visible=True)),
    hovermode='x unified',
)
fig.show()
fig.write_html('interactive_line.html')
# 2. Interactive bar chart
df_2007 = gapminder.query('year == 2007 and continent == "Asia"')
df_top10 = df_2007.nlargest(10, 'gdpPercap')

fig2 = px.bar(df_top10, x='country', y='gdpPercap', color='lifeExp',
              title='Top 10 Asian Countries by GDP (2007)',
              color_continuous_scale='Viridis')

# Customize axis labels
fig2.update_layout(xaxis_tickangle=-45)
fig2.show()
fig2.write_html('interactive_bar.html')
```

# Task 4.3:  Plotly Heatmap & Sunburst

```python
# Load dataset
df = pd.read_excel('Practice_Dataset.xlsx')

# 1. Correlation Heatmap
# Select numeric columns only
numeric_df = df.select_dtypes(include=[np.number])
corr = numeric_df.corr()

# Create interactive heatmap
fig = px.imshow(corr, text_auto='.2f', color_continuous_scale='RdBu_r',
                title='Interactive Correlation Heatmap')
fig.show()
fig.write_html('plotly_heatmap.html')

# 2. Sunburst Chart
# Hierarchical data visualization
gapminder_2007 = px.data.gapminder().query('year == 2007')
fig2 = px.sunburst(gapminder_2007, path=['continent', 'country'],
                   values='pop', color='lifeExp',
                   color_continuous_scale='RdYlGn',
                   title='World Population by Continent & Country')
fig2.show()
fig2.write_html('sunburst.html')
```

# Task 4.4:  Build a Mini Dashboard with Subplots

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go

# Create a 2x2 dashboard grid
fig = make_subplots(rows=2, cols=2,
    subplot_titles=('Revenue Trend', 'Sales by Category', 
                    'Score Distribution', 'Regional Comparison'),
    specs=[[{'type': 'scatter'}, {'type': 'bar'}],
           [{'type': 'histogram'}, {'type': 'pie'}]])

# 1. Line chart
fig.add_trace(go.Scatter(x=list(range(1, 13)), y=[120,135,148,162,155,170,185,178,195,210,225,240],
    mode='lines+markers', name='Revenue', line=dict(color='#0D9488')), row=1, col=1)

# 2. Bar chart
fig.add_trace(go.Bar(x=['Electronics','Clothing','Food','Books'],
    y=[450,320,280,150], name='Sales',
    marker_color=['#0D9488','#1B2A4A','#F59E0B','#EF4444']), row=1, col=2)

# 3. Histogram
fig.add_trace(go.Histogram(x=np.random.normal(75, 15, 500),
    name='Scores', marker_color='#0D9488'), row=2, col=1)

# 4. Pie chart
fig.add_trace(go.Pie(labels=['East','West','North','South'],
    values=[40,30,20,10], name='Region'), row=2, col=2)

# Update layout and save
fig.update_layout(height=700, title_text='Sales Dashboard', showlegend=False)
fig.show()
fig.write_html('mini_dashboard.html') # Saved as HTML file [cite: 115]
```
